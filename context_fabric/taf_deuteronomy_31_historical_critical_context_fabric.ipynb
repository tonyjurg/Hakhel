# 1 - Introduction<a class="anchor" id="bullet1"></a>
##### [Back to ToC](#TOC)

## 1.1 - General introduction<a class="anchor" id="bullet1x1"></a>

This Jupyter Notebook investigates a seemingly simple question: why does Deuteronomy 31:12 explicitly include `הטף` — often translated as “the little ones” or “the children” — among those summoned to hear the Torah at Sukkot in every seventh year?

The starting point is the plain text of Deuteronomy 31:10–13. The passage gathers “the people: the men, the women, the `טף`, and your resident alien who is within your gates,” so that they may hear, learn, fear YHWH, and observe the Torah. Verse 13 then turns separately to children “who have not known.” That wording invites lexical, literary, and historical questions which should not be collapsed into one another.

All corpus retrieval in this notebook goes through a *local Context Fabric MCP server* and the local ETCBC BHSA 2021 Text-Fabric files (created by notebook [`shebanq_public_queries.ipynb`](../SHEBANQ-BHSA/shebanq_public_queries.ipynb)).

## 1.2 - Reading guide: keep three evidential layers separate<a class="anchor" id="bullet1"></a>
##### [Back to ToC](#TOC)

A digital corpus can answer some questions with precision and cannot answer others by itself. This notebook therefore marks three layers:

1. **Corpus observation** — what BHSA encodes: words, lexemes, morphology, syntax, references, and distributions.
2. **Literary observation** — how Deuteronomy 31:10–13 relates to nearby Deuteronomic formulas and to Joshua 8:35.
3. **Historical inference** — what these patterns may suggest about scribal ideology, covenant renewal, collective identity, pedagogy, and redaction.

The first layer is reproducible by query. The second depends on close reading. The third is probabilistic.



## 1.3 - Working hypotheses<a class="anchor" id="bullet1x3"></a>
##### [Back to ToC](#TOC)

- `טף` is a collective social category, not a technical label for one exact age.
- Across the Hebrew Bible it frequently denotes dependents or noncombatants in household movement, war, capture, provision, and danger.
- Its inclusion in Deuteronomy 31:12 transfers that vulnerable/dependent category into the covenant assembly, making “all the people” concrete rather than just rhetorical.
- Deuteronomy 29:10, Deuteronomy 31:12, and Joshua 8:35 form a particularly important inclusion formula because `טף` and `גר` occur together only in these three verses.
- The passage is best read as a program of intergenerational covenant reproduction. Whether it describes a regularly performed pre-exilic institution is a separate and much harder historical question.

# 2 - Two types of access: direct corpus access and MCP-mediated access<a class="anchor" id="bullet2"></a>
##### [Back to ToC](#TOC)



The phrase “accessing BHSA” can describe at least two technically different routes. They consult the *same dataset*, but the route from research code to corpus differs. The following table compares a direct  and the MCP mediated access method.

| Question | Direct library/file access | MCP-mediated access |
|---|---|---|
| Who opens BHSA? | The notebook's Python process opens the local Text-Fabric files through Text-Fabric, a file format specific Python library. | The local Context Fabric MCP server opens the local Text-Fabric files. |
| How are requests expressed? | Python method calls tied to the [Text-Fabric API](https://annotation.github.io/text-fabric/tf/). | Standardized [MCP tool calls](https://context-fabric.ai/docs/mcp/guides/tools) such as `describe_corpus`, `search`, and `get_node_features`. |
| Where does corpus state live? | Inside the notebook kernel. | Inside the separately running local MCP server. |
| Main strength | Maximum flexibility for custom, low-level corpus programming. | Tool discovery, explicit schemas, client/language independence, and a clean separation between corpus service and analysis. |
| Main trade-off | Research code is closely coupled to the Text-fabric API and loads the corpus itself. | Queries must fit the tools and schemas exposed by the server; a server process must be running. |



## 2.1 - Which access type does this investigation use?<a class="anchor" id="bullet2x1"></a>
##### [Back to ToC](#TOC)

All textual retrieval, feature inspection, concordance searching, passage retrieval, and syntactic analysis in this notebook will use the MCP-mediated access. The main reason is not technical, but methodical: all textual domains consulted in this investigation will use MCP-mediated access.

The notebook talks to a Context Fabric MCP server running on this same computer, and that server consults only the local ETCBC BHSA 2021 files.

It does not parse the biblical text or calculate the reported linguistic results directly from `.tf` files.


## 2.2 - Do stdio and SSE represent two more datasets?<a class="anchor" id="bullet2x2"></a>
##### [Back to ToC](#TOC)

No. They are two ways to transport MCP messages:

- stdio transport: the client launches `cfabric-mcp` as a child process and communicates over standard input/output. It needs no listening port and is convenient for a single desktop client.

- local network transport: the server runs independently and listens on a loopback port. Context Fabric supports SSE and streamable HTTP variants. This is convenient for a notebook because the server can stay loaded while many cells open sessions.

This notebook uses local SSE at `http://127.0.0.1:8765/sse`. The endpoint is bound to loopback, so it is not a hosted service. Switching from SSE to stdio or streamable HTTP would change the connection mechanism, not the corpus: the dataset would still be BHSA 2021.

```text
Direct access                         Access used in this notebook

notebook                             notebook
   │ Python corpus API                  │ MCP tool calls over local SSE
   ▼                                    ▼
local BHSA files                     local Context Fabric MCP server
                                           │ Context Fabric corpus access
                                           ▼
                                       local BHSA files
```

# 5 - Check that the MCP server is local<a class="anchor" id="bullet5"></a>
##### [Back to ToC](#TOC)

The notebook first probes `127.0.0.1:8765`. If no process is listening, it launches the repo-local `cfabric-mcp.exe` with the repo-local BHSA path.

The feature list is broad enough for this study: references, Hebrew forms, lexemes, glosses, morphology, phrase/clause functions and types, discourse domains, and parent relations. Context Fabric compiles a `.cfm` cache beside the corpus on first use. That first compilation is resource-intensive; later launches reuse the cache.

Binding to `127.0.0.1` rather than `0.0.0.0` keeps the research service accessible only from this computer.

Note that the first time this cell is runned, it might take a long time since it needs to download and compile many BHSA related data files. 

````python
# Start and verify a genuinely local Context Fabric MCP server

FEATURES = " ".join(
    [
        "book", "chapter", "verse",
        "lex", "lex_utf8", "g_lex_utf8", "g_word_utf8", "g_cons_utf8",
        "voc_lex_utf8", "gloss",
        "sp", "pdp", "gn", "nu", "ps", "st", "vt", "vs", "ls",
        "language", "function", "typ", "domain", "kind", "rela",
        "mother", "mother_object_type", "label", "trailer_utf8", "freq_lex",
    ]
)


def port_is_open(host: str, port: int, timeout: float = 0.5) -> bool:
    """Return True when a TCP connection can be made to host:port."""
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.settimeout(timeout)
        return sock.connect_ex((host, port)) == 0


def read_log(path: Path) -> str:
    """Read a log file without failing on encoding problems."""
    if not path.exists():
        return "[log file not created]"

    content = path.read_text(
        encoding="utf-8",
        errors="replace",
    )

    return content if content else "[empty]"


SERVER_DIR.mkdir(parents=True, exist_ok=True)

stdout_path = SERVER_DIR / "cfabric-mcp.stdout.log"
stderr_path = SERVER_DIR / "cfabric-mcp.stderr.log"
pid_path = SERVER_DIR / "cfabric-mcp.pid"

# Run the server with the repository-local Python environment.
venv_python = VENV_PYTHON

if not venv_python.is_file():
    raise RuntimeError(
        "The repository-local Python interpreter was not found:\n"
        f"{venv_python}\n\n"
    )

if not CORPUS_DIR.is_dir():
    raise RuntimeError(
        "The Context Fabric corpus directory was not found:\n"
        f"{CORPUS_DIR}"
    )


# Verify that both distributions are installed in the MCP environment.
package_probe_code = """
from importlib.metadata import version

for package in ("cfabric-mcp", "context-fabric"):
    print(f"{package}={version(package)}")
"""

package_probe = subprocess.run(
    [
        str(venv_python),
        "-c",
        package_probe_code,
    ],
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)

if package_probe.returncode != 0:
    repair_command = subprocess.list2cmdline(
        [
            str(venv_python),
            "-m",
            "pip",
            "install",
            "-r",
            str(CONTEXT_DIR / "requirements.txt"),
        ]
    )

    raise RuntimeError(
        "The dedicated MCP environment could not load its packages.\n\n"
        f"Python: {venv_python}\n"
        f"Exit code: {package_probe.returncode}\n\n"
        f"STDOUT:\n{package_probe.stdout or '[empty]'}\n\n"
        f"STDERR:\n{package_probe.stderr or '[empty]'}\n\n"
        "Repair the environment by running:\n"
        f"{repair_command}"
    )

print("Dedicated MCP environment:")
print(package_probe.stdout.strip())


# Load the cfabric-mcp console entry point through the virtual environment's
# Python interpreter. This bypasses a potentially stale Windows .exe launcher.
console_entry_code = """
from importlib.metadata import distribution

entry_point = next(
    ep
    for ep in distribution("cfabric-mcp").entry_points
    if ep.group == "console_scripts" and ep.name == "cfabric-mcp"
)

raise SystemExit(entry_point.load()())
"""


server_started_by_notebook = False
server_process = None

if port_is_open(MCP_HOST, MCP_PORT):
    print(
        f"A service is already listening on "
        f"{MCP_HOST}:{MCP_PORT}; no new server was started."
    )

else:
    server_command = [
        str(venv_python),
        "-c",
        console_entry_code,
        "--corpus", f"bhsa={CORPUS_DIR}",
        "--features", FEATURES,
        "--sse", str(MCP_PORT),
        "--host", MCP_HOST,
        "--verbose",
    ]

    # Clear logs from previous runs.
    stdout_path.write_text("", encoding="utf-8")
    stderr_path.write_text("", encoding="utf-8")

    stdout_handle = stdout_path.open("w", encoding="utf-8")
    stderr_handle = stderr_path.open("w", encoding="utf-8")

    creationflags = getattr(subprocess, "CREATE_NO_WINDOW", 0)

    try:
        server_process = subprocess.Popen(
            server_command,
            cwd=str(CORPUS_DIR),
            stdout=stdout_handle,
            stderr=stderr_handle,
            creationflags=creationflags,
        )
    finally:
        # The child process retains its own log handles.
        stdout_handle.close()
        stderr_handle.close()

    pid_path.write_text(
        str(server_process.pid),
        encoding="ascii",
    )

    server_started_by_notebook = True
    startup_timeout_seconds = 120
    deadline = time.time() + startup_timeout_seconds

    while time.time() < deadline:
        if port_is_open(MCP_HOST, MCP_PORT):
            break

        return_code = server_process.poll()

        if return_code is not None:
            raise RuntimeError(
                "Local Context Fabric MCP exited during startup.\n\n"
                f"Exit code: {return_code}\n"
                f"Python: {venv_python}\n"
                f"Corpus: {CORPUS_DIR}\n"
                f"Host: {MCP_HOST}\n"
                f"Port: {MCP_PORT}\n\n"
                f"STDOUT:\n{read_log(stdout_path)}\n\n"
                f"STDERR:\n{read_log(stderr_path)}"
            )

        time.sleep(2)

    if not port_is_open(MCP_HOST, MCP_PORT):
        # Stop only the process started by this notebook cell.
        if server_process.poll() is None:
            server_process.terminate()

            try:
                server_process.wait(timeout=10)
            except subprocess.TimeoutExpired:
                server_process.kill()
                server_process.wait(timeout=10)

        raise RuntimeError(
            "Local Context Fabric MCP did not open its port within "
            f"{startup_timeout_seconds} seconds.\n\n"
            f"Process exit code: {server_process.returncode}\n"
            f"Python: {venv_python}\n"
            f"Corpus: {CORPUS_DIR}\n"
            f"Host: {MCP_HOST}\n"
            f"Port: {MCP_PORT}\n\n"
            f"STDOUT:\n{read_log(stdout_path)}\n\n"
            f"STDERR:\n{read_log(stderr_path)}"
        )

    print(
        f"Started local Context Fabric MCP at "
        f"http://{MCP_HOST}:{MCP_PORT}/sse "
        f"(PID {server_process.pid})."
    )
````

# 6 - Direct MCP initialization and tool discovery<a class="anchor" id="bullet6"></a>
##### [Back to ToC](#TOC)

This is the investigation's first corpus interaction. We do not assume tool names copied from documentation. We initialize an MCP session and ask the running local server to describe itself and list its callable tools with JSON schemas.

This avoids a surprisingly easy reproducibility error: writing code for an imagined `bhsa_search` tool when the actual server exposes `search`, `get_passages`, `get_node_features`, and discovery tools.

````python
# Direct MCP initialization and tool discovery
from mcp import ClientSession
from mcp.client.sse import sse_client


# Decode MCP content blocks
def decode_mcp_result(result: Any) -> Any:
    decoded = []
    for item in result.content:
        text = getattr(item, "text", str(item))
        try:
            decoded.append(json.loads(text))
        except json.JSONDecodeError:
            decoded.append(text)
    return decoded[0] if len(decoded) == 1 else decoded


# Call one MCP tool
async def mcp_call(tool_name: str, arguments: dict[str, Any] | None = None) -> Any:
    async with sse_client(MCP_SSE_URL) as (read, write):
        async with ClientSession(read, write) as session:
            await session.initialize()
            result = await session.call_tool(tool_name, arguments or {})
            return decode_mcp_result(result)


# Read the local server description
async def discover_local_server() -> dict[str, Any]:
    async with sse_client(MCP_SSE_URL) as (read, write):
        async with ClientSession(read, write) as session:
            initialized = await session.initialize()
            listed = await session.list_tools()
            return {
                "server": initialized.serverInfo.model_dump(),
                "protocol_version": initialized.protocolVersion,
                "tools": [tool.model_dump() for tool in listed.tools],
            }


discovery = await discover_local_server()
tool_catalog = pd.DataFrame(
    [
        {
            "tool": tool["name"],
            "description": tool.get("description", "").splitlines()[0],
            "required_arguments": tool.get("inputSchema", {}).get("required", []),
        }
        for tool in discovery["tools"]
    ]
)
display(pd.Series(discovery["server"], name="server value").to_frame())
display(tool_catalog)

(RESULTS_DIR / "mcp_tool_inventory.json").write_text(
    json.dumps(discovery, ensure_ascii=False, indent=2) + "\n",
    encoding="utf-8",
)
````

# 7 - Let MCP describe the BHSA corpus and relevant features<a class="anchor" id="bullet7"></a>
##### [Back to ToC](#TOC)

Before searching, we ask the server:

- which corpus is loaded;
- which node types and section levels exist;
- what `lex_utf8`, `g_word_utf8`, `gloss`, `sp`, `function`, `typ`, and `domain` mean in this dataset;
- what search syntax this server accepts.

This step helps us avoid mixing conventions. For example, BHSA stores Deuteronomy's `book` feature as `Deuteronomium`, although MCP renders human-facing references as `Deuteronomy`.

````python
# Let MCP describe the BHSA corpus and relevant features
corpora_info = await mcp_call("list_corpora")
corpus_description = await mcp_call("describe_corpus", {"corpus": "bhsa"})
feature_description = await mcp_call(
    "describe_feature",
    {
        "corpus": "bhsa",
        "feature": [
            "book", "chapter", "verse", "lex_utf8", "g_word_utf8",
            "gloss", "sp", "function", "typ", "domain",
        ],
        "sample_limit": 12,
    },
)
search_guide = await mcp_call("search_syntax_guide")

display(pd.Series(corpora_info, name="value").to_frame())
display(pd.DataFrame(corpus_description["node_types"]))

feature_rows = []
for name, details in feature_description["features"].items():
    feature_rows.append(
        {
            "feature": name,
            "kind": details["kind"],
            "value_type": details["value_type"],
            "node_types": ", ".join(details["node_types"]),
            "description": details["description"],
            "unique_values": details["unique_values"],
        }
    )
display(pd.DataFrame(feature_rows))
display(pd.Series(search_guide, name="value").to_frame())
````

# 8 - Retrieve the primary passages directly from local BHSA<a class="anchor" id="bullet8"></a>
##### [Back to ToC](#TOC)

We begin with Deuteronomy 31:10–13, rather than verse 12 in isolation:

- verse 10 fixes the seventh-year/Sukkot schedule;
- verse 11 commands public Torah reading at the chosen place;
- verse 12 defines the audience and purpose;
- verse 13 explains intergenerational continuation through children “who have not known.”

Two comparison texts are added:

- **Deuteronomy 29:9–11**: covenant standing before YHWH, including `טף`, women, and resident alien;
- **Joshua 8:34–35**: narrative Torah reading before the assembly, including women, `טף`, and resident alien.

Joshua 8:35 is not independent proof that Deuteronomy 31 was historically performed exactly as written. Literarily, however, it is a powerful Deuteronomistic realization of the same totalizing audience ideal.

The installed Context-Fabric MCP version currently returns an internal `Computeds.sections` error from its convenience tool `get_passages`. We therefore retrieve the same verses with exact `search` templates, the same route used for the investigation below. The small name mapping is stated in the code: BHSA stores `Deuteronomium` and `Josua` in its `book` feature, while Context Fabric renders the references in English as `Deuteronomy` and `Joshua`. No verse text is supplied by this notebook.


````python
# Retrieve the selected verses through exact Context-Fabric searches.
# Each tuple records: displayed English book, BHSA book-feature value, chapter, verse.
PRIMARY_SECTIONS = [
    ("Deuteronomy", "Deuteronomium", 29, 9),
    ("Deuteronomy", "Deuteronomium", 29, 10),
    ("Deuteronomy", "Deuteronomium", 29, 11),
    ("Deuteronomy", "Deuteronomium", 31, 10),
    ("Deuteronomy", "Deuteronomium", 31, 11),
    ("Deuteronomy", "Deuteronomium", 31, 12),
    ("Deuteronomy", "Deuteronomium", 31, 13),
    ("Joshua", "Josua", 8, 34),
    ("Joshua", "Josua", 8, 35),
]

primary_rows = []
for display_book, bhsa_book, chapter, verse in PRIMARY_SECTIONS:
    response = await mcp_call(
        "search",
        {
            "corpus": "bhsa",
            "template": f"v:verse book={bhsa_book} chapter={chapter} verse={verse}",
            "return_type": "passages",
            "limit": 2,
        },
    )
    expected_reference = f"{display_book} {chapter}:{verse}"
    assert response["total_count"] == 1, (
        f"Expected one BHSA verse for {expected_reference}; "
        f"Context Fabric returned {response['total_count']}."
    )
    passage = response["passages"][0]
    assert passage["reference"] == expected_reference
    primary_rows.append(
        {
            "section_ref": passage["reference"],
            "node": passage["node"],
            "text": passage["text"],
        }
    )

primary_df = pd.DataFrame(primary_rows)
display(primary_df)
````

# 9 - Build Unicode-safe Hebrew search terms<a class="anchor" id="bullet9"></a>
##### [Back to ToC](#TOC)

On Windows PowerShell 5, literal Hebrew piped into a Python process may be converted to question marks before MCP receives it. A notebook normally stores Unicode safely, but explicit code-point construction keeps the query safe across the different environments used here.

Important BHSA lexemes:

- `טף` — collective dependents / little ones;
- `אישׁ`, `אשׁה`, `גר` — man, woman, resident alien/sojourner;
- `שׁמע`, `למד`, `ירא`, `שׁמר`, `עשׂה` — hear, learn, fear, keep, do;
- `בן`, `ידע` — child/son and know in verse 13.

````python
# Build Unicode-safe Hebrew search terms
H = {
    "taf": "\u05d8\u05e3",
    "man": "\u05d0\u05d9\u05e9\u05c1",
    "woman": "\u05d0\u05e9\u05c1\u05d4",
    "ger": "\u05d2\u05e8",
    "hear": "\u05e9\u05c1\u05de\u05e2",
    "learn": "\u05dc\u05de\u05d3",
    "fear": "\u05d9\u05e8\u05d0",
    "keep": "\u05e9\u05c1\u05de\u05e8",
    "do": "\u05e2\u05e9\u05c2\u05d4",
    "son": "\u05d1\u05df",
    "know": "\u05d9\u05d3\u05e2",
}
pd.Series(H, name="BHSA lex_utf8").to_frame()
````

# 10 - Verify the audience formula and pedagogical verb chain<a class="anchor" id="bullet10"></a>
##### [Back to ToC](#TOC)

The search captures the verse plus each relevant word. Same-level query lines follow canonical order in Context Fabric's template language.

Two grammatical observations matter:

1. The audience categories occur inside the object of the singular imperative “assemble.”
2. The subsequent verbs are third-person plural: “they may hear … learn … fear … keep … do.” At the grammatical level, the stated purpose belongs to the assembled collective. The text does not mark `טף` as a parenthetical group exempt from the purpose chain.

Verse 13 then narrows attention to descendants “who have not known,” supplying an explicit intergenerational rationale without defining a minimum age.

````python
# Verify the audience formula and pedagogical verb chain
audience_template = f'''verse book=Deuteronomium chapter=31 verse=12
  m:word lex_utf8={H['man']}
  w:word lex_utf8={H['woman']}
  t:word lex_utf8={H['taf']}
  g:word lex_utf8={H['ger']}'''

purpose_template = f'''verse book=Deuteronomium chapter=31 verse=12
  h:word lex_utf8={H['hear']}
  l:word lex_utf8={H['learn']}
  f:word lex_utf8={H['fear']}
  k:word lex_utf8={H['keep']}
  d:word lex_utf8={H['do']}'''

verse13_template = f'''verse book=Deuteronomium chapter=31 verse=13
  b:word lex_utf8={H['son']}
  k:word lex_utf8={H['know']}
  h:word lex_utf8={H['hear']}
  l:word lex_utf8={H['learn']}
  f:word lex_utf8={H['fear']}'''

audience_result = await mcp_call(
    "search", {"template": audience_template, "return_type": "results", "limit": 20, "corpus": "bhsa"}
)
purpose_result = await mcp_call(
    "search", {"template": purpose_template, "return_type": "results", "limit": 20, "corpus": "bhsa"}
)
verse13_result = await mcp_call(
    "search", {"template": verse13_template, "return_type": "results", "limit": 20, "corpus": "bhsa"}
)


# Collect the named nodes from the first match
def captured_nodes(result: dict[str, Any]) -> pd.DataFrame:
    return pd.DataFrame(result["results"][0])


display(Markdown("### Deuteronomy 31:12 audience"))
display(captured_nodes(audience_result))
display(Markdown("### Deuteronomy 31:12 purpose chain"))
display(captured_nodes(purpose_result))
display(Markdown("### Deuteronomy 31:13 intergenerational chain"))
display(captured_nodes(verse13_result))
````

# 11 - Inspect morphology and syntax, not only strings<a class="anchor" id="bullet11"></a>
##### [Back to ToC](#TOC)

We now ask MCP for features on the captured word nodes and for the phrase/clause that contain `טף`.

This matters because a translation such as “little ones” may suggest an age-specific plural noun. BHSA instead analyzes `טף` as:

- a substantive;
- masculine singular;
- functioning collectively;
- glossed by the annotators as “those unable to march.”

That gloss should not be treated as inspired lexicography. It is, however, a compact summary of the distributional fact that the term repeatedly contrasts mobile/fighting adults with dependents.

````python
# Inspect morphology and syntax, not only strings
captured_word_nodes = [
    item["node"]
    for payload in (audience_result, purpose_result, verse13_result)
    for item in payload["results"][0]
    if item["otype"] == "word"
]
word_features = await mcp_call(
    "get_node_features",
    {
        "nodes": sorted(set(captured_word_nodes)),
        "features": [
            "lex_utf8", "g_word_utf8", "gloss", "sp", "pdp",
            "gn", "nu", "ps", "st", "vt", "vs", "language",
        ],
        "corpus": "bhsa",
    },
)
word_feature_df = pd.DataFrame(word_features["nodes"])
display(word_feature_df)

deut_structure = await mcp_call(
    "search",
    {
        "template": f'''verse book=Deuteronomium chapter=31 verse=12
  c:clause
    p:phrase
      w:word lex_utf8={H['taf']}''',
        "return_type": "results",
        "limit": 20,
        "corpus": "bhsa",
    },
)
josh_structure = await mcp_call(
    "search",
    {
        "template": f'''verse book=Josua chapter=8 verse=35
  c:clause
    p:phrase
      w:word lex_utf8={H['taf']}''',
        "return_type": "results",
        "limit": 20,
        "corpus": "bhsa",
    },
)

structure_nodes = [
    item["node"]
    for payload in (deut_structure, josh_structure)
    for item in payload["results"][0]
]
structure_features = await mcp_call(
    "get_node_features",
    {
        "nodes": structure_nodes,
        "features": ["function", "typ", "domain", "kind", "rela", "lex_utf8", "g_word_utf8", "gloss"],
        "corpus": "bhsa",
    },
)
structure_df = pd.DataFrame(structure_features["nodes"])
display(structure_df)
````

## 11.1 - Interpretation of the local syntax<a class="anchor" id="bullet11x1"></a>
##### [Back to ToC](#TOC)

BHSA analyzes the opening command of Deuteronomy 31:12 as a quotation-domain verbal clause (`ZIm0`, an imperative pattern). The entire sequence “the people—the men, the women, the `טף`, and your resident alien” belongs to one object phrase. `טף` is therefore structurally integrated into the commanded assembly.

Joshua 8:35 differs syntactically because it narrates reading “before” the whole assembly; BHSA labels the phrase containing women, `טף`, and resident alien as a complement. The social inclusion formula survives even though the clause's grammatical architecture changes.

This difference is worth noting. Joshua does not mechanically copy a complete Deuteronomic sentence, but reuses the same constituency formula in a new narrative syntax.

# 12 - Retrieve every `טף` occurrence as a full verse<a class="anchor" id="bullet12"></a>
##### [Back to ToC](#TOC)

A lexeme-level study should not infer meaning from a dictionary gloss alone. We search the entire corpus for a verse containing a `טף` word, capture both the verse and the word node, and inspect all 42 contexts.

Using the verse as the query root is important. A query rooted at `word` returns only the matched word as its “passage,” which hides the social frame.

````python
# Retrieve every `טף` occurrence as a full verse
all_taf_template = f'''v:verse
  w:word lex_utf8={H['taf']}'''

all_taf_result = await mcp_call(
    "search",
    {
        "template": all_taf_template,
        "return_type": "results",
        "limit": 100,
        "max_override": True,
        "corpus": "bhsa",
    },
)
assert all_taf_result["total_count"] == 42

occurrence_rows = []
for match in all_taf_result["results"]:
    verse_item = next(item for item in match if item["otype"] == "verse")
    word_item = next(item for item in match if item["otype"] == "word")
    occurrence_rows.append(
        {
            "reference": verse_item["section_ref"],
            "verse_node": verse_item["node"],
            "word_node": word_item["node"],
            "matched_form": word_item["text"].strip(),
            "verse_text": verse_item["text"].strip(),
        }
    )

taf_df = pd.DataFrame(occurrence_rows)
display(taf_df[["reference", "matched_form", "verse_text"]])
````

# 13 - Classify the contexts<a class="anchor" id="bullet13"></a>
##### [Back to ToC](#TOC)

Context Fabric retrieves the occurrences, references, Hebrew forms, verse text, morphology, and syntax. It does **not** supply the contextual labels used below. The classification is a separate act of scholarly coding based on close reading of the 42 verses.

The assignments are stored in `context_classification_mapping.json`. They are researcher-supplied interpretations, not BHSA features or values returned by MCP. Keeping them outside the code makes changes to the scholarly coding visible in an ordinary diff.

A fresh verse-by-verse review showed that the former label *movement, settlement, or dependent care* was too broad. It combined journeys and protected residence with the materially different question of feeding a household. Every reference is now assigned to one of four broad frames:

- **conflict, captivity, or lethal judgment**;
- **household movement, displacement, or protected residence**;
- **household provision or survival**;
- **covenant, cultic, or administrative inclusion**.

The fourth frame is worth adding because Genesis 43:8; 47:12, 24; and 50:21 concern continued life, food, or provision rather than the location of the household. The distinction is small enough to remain readable and large enough to prevent four materially different verses from disappearing inside the largest group. Adding further broad frames would begin to produce one-verse categories. A narrower `context_detail` therefore records the immediate event without turning every detail into a separate count.

The mapping also gives the relevant unpointed Hebrew phrase, a contextual rendering of `טף`, a close working translation, and a brief reason for the assignment. The translations are aids to reading the Hebrew; they are not quotations from a published Bible. *Little ones* remains the normal rendering. *Household dependants* is used where `טף` functions as a wider household aggregate, while Numbers 31:17–18 and 2 Chronicles 20:13; 31:18 require still more specific wording.

The procedure is fixed when the notebook runs, but interpretive in its construction. The validation requires every retrieved reference to occur exactly once, every frame to have a definition, and every mapped Hebrew phrase to be present in its BHSA verse after pointing and cantillation are removed. An omitted, duplicated, unexpected, or textually mismatched reference stops the notebook before totals are calculated.

Two borderline decisions should remain visible. Jeremiah 41:16 belongs with conflict and captivity because the little ones are among people recovered from Ishmael, not simply a travelling group. In 2 Chronicles 31:18, `להתיחש` denotes genealogical enrollment; the verse therefore belongs to institutional inclusion, but not to public assembly in the narrower sense.


````python
# Apply the explicit researcher-supplied contextual mapping
context_mapping = json.loads(CONTEXT_MAPPING_FILE.read_text(encoding="utf-8"))

assert context_mapping["mapping_type"] == "researcher-supplied interpretive classification"
assert context_mapping["input_key"] == "reference"

mapping_df = pd.DataFrame(context_mapping["occurrences"])
expected_mapping_columns = ["reference", *context_mapping["output_fields"]]
assert list(mapping_df.columns) == expected_mapping_columns

# No reference may be absent from the mapping or assigned more than once.
retrieved_references = set(taf_df["reference"])
mapped_references = set(mapping_df["reference"])
duplicate_codings = sorted(
    mapping_df.loc[mapping_df["reference"].duplicated(keep=False), "reference"].unique()
)
unclassified_references = retrieved_references - mapped_references
unexpected_references = mapped_references - retrieved_references
unexpected_frames = set(mapping_df["context_frame"]) - set(context_mapping["frame_definitions"])

assert not duplicate_codings, f"References assigned more than once: {duplicate_codings}"
assert not unclassified_references, f"Retrieved references without a frame: {sorted(unclassified_references)}"
assert not unexpected_references, f"Mapped references absent from retrieval: {sorted(unexpected_references)}"
assert not unexpected_frames, f"Frames without a definition: {sorted(unexpected_frames)}"

# Compare consonants so that the readable, unpointed phrases can be checked
# against the pointed and cantillated BHSA verse text.
def consonantal_hebrew(value: str) -> str:
    decomposed = unicodedata.normalize("NFD", value)
    return "".join(
        character
        for character in decomposed
        if unicodedata.category(character).startswith("L")
    )

phrase_check_df = taf_df[["reference", "verse_text"]].merge(
    mapping_df[["reference", "relevant_hebrew_phrase"]],
    on="reference",
    how="inner",
    validate="one_to_one",
)
phrase_mismatches = [
    row.reference
    for row in phrase_check_df.itertuples()
    if consonantal_hebrew(row.relevant_hebrew_phrase)
    not in consonantal_hebrew(row.verse_text)
]
assert not phrase_mismatches, f"Mapped Hebrew phrase not found in BHSA verse: {phrase_mismatches}"

taf_df = taf_df.merge(
    mapping_df,
    on="reference",
    how="left",
    validate="one_to_one",
)

context_counts = (
    taf_df["context_frame"]
    .value_counts()
    .rename_axis("context_frame")
    .reset_index(name="occurrences")
)
context_counts["percent"] = (
    100 * context_counts["occurrences"] / len(taf_df)
).round(1)

display(pd.Series({
    "mapping_type": context_mapping["mapping_type"],
    "mapping_file": CONTEXT_MAPPING_FILE.relative_to(REPO_ROOT).as_posix(),
    "mapping_sha256": context_mapping_sha256,
}))
display(context_counts)
classification_review_columns = [
    "reference",
    "matched_form",
    "relevant_hebrew_phrase",
    "taph_rendering",
    "working_translation",
    "context_frame",
    "context_detail",
    "classification_note",
]
classification_review_df = taf_df[classification_review_columns].copy()
display(classification_review_df)

taf_df.to_csv(
    RESULTS_DIR / "taf_occurrences_context_classified.csv",
    index=False,
    encoding="utf-8-sig",
)
classification_review_df.to_csv(
    RESULTS_DIR / "taf_context_classification_review.csv",
    index=False,
    encoding="utf-8-sig",
)
````

# 16 - Historical-critical synthesis<a class="anchor" id="bullet16"></a>
##### [Back to ToC](#TOC)

## 16.1 - Lexical result: dependent collective, not an exact age bracket<a class="anchor" id="bullet16x1"></a>
##### [Back to ToC](#TOC)

BHSA finds 42 instances of `טף`. It analyzes every occurrence as masculine singular, even when English idiom requires a plural. The forms may carry pronominal suffixes—“their `טף`,” “your `טף`,” “our `טף`”—which reinforces the household/dependency relation.

The contexts repeatedly oppose or coordinate `טף` with adults able to fight, travel independently, negotiate, or hold responsibility. “Little ones” is defensible, but “dependents” or “young dependents” often better preserves the social function. The corpus does **not** give an age threshold. Numbers 31:17–18 can subdivide the category by sex, and other passages coordinate `טף` with `בנים`, showing that the semantic boundaries are contextual rather than a modern census category.

This contextual flexibility is not a new meaning inferred from the four frames. Brown–Driver–Briggs describes `טף` as a masculine collective for children and notes several household passages in which women are included or implied. Athalya Brenner likewise calls it a non-gendered collective noun. These lexical observations justify varying the English gloss, but they do not assign the contextual frames; those assignments still depend on the syntax and episode of each verse.

## 16.2 - Rhetorical result: `כל־העם` is unpacked<a class="anchor" id="bullet16x2"></a>
##### [Back to ToC](#TOC)

Deuteronomy 31:12 first says “assemble the people” and then appositionally specifies men, women, `טף`, and resident alien. The list prevents an elite or military reading of “the people.” It makes the covenant public include:

- adult Israelite men;
- adult Israelite women;
- dependent/non-autonomous household members;
- resident outsiders incorporated into local Israelite life.

The `טף` are not merely present because parents need childcare. At the level of the sentence, they belong to the object of the assembly command, and the plural purpose chain is predicated of the assembled public.

## 16.3 - Pedagogical result: hearing precedes durable obedience<a class="anchor" id="bullet16x3"></a>
##### [Back to ToC](#TOC)

Verse 12 moves from hearing and learning to fearing YHWH, keeping, and doing the Torah. Verse 13 repeats hearing/learning/fearing for “their children who have not known.” The ceremony therefore imagines covenant knowledge as periodically renewed, orally transmitted, embodied in attendance, and extended across generations.

This does not require every infant to comprehend a legal recitation proposition by proposition. The text works at a collective level. Different hearers may participate with different cognitive capacities while belonging to the same covenant event.

## 16.4 - Literary result: Deuteronomy 29 → Deuteronomy 31 → Joshua 8<a class="anchor" id="bullet16x4"></a>
##### [Back to ToC](#TOC)

The rare `טף + גר` co-occurrence links three scenes:

- covenant standing (Deut 29);
- commanded periodic reading (Deut 31);
- narrated reading and covenant ceremony (Josh 8).

This sequence fits a Deuteronomic/Deuteronomistic program in which written Torah becomes publicly authoritative through recitation before a totalized community. Joshua 8:35 functions literarily as a realization of Mosaic Torah ideology, whether or not it preserves an independent archival report of an early ceremony.

## 16.5 - Social-historical result: a counterfactual public<a class="anchor" id="bullet16x5"></a>
##### [Back to ToC](#TOC)

In ordinary `טף` contexts, dependents are carried, left in protected settlements, fed, captured, threatened, or killed. Deuteronomy 31 converts that social dependence into ritual membership. Those least likely to be counted as autonomous political actors are nonetheless required for the covenant public to be complete.

“Counterfactual” is useful here: the assembly imagines Israel not as the subset able to march, own, fight, or read, but as everyone who must live under and transmit the Torah.

## 16.6 - Compositional and institutional questions<a class="anchor" id="bullet16x6"></a>
##### [Back to ToC](#TOC)

Historical-critical scholarship commonly relates Deuteronomy's written-Torah ideology, central sanctuary, covenant rhetoric, and public instruction to late-monarchic and exilic scribal formation, often in conversation with the Josianic reform narrative and later Deuteronomistic editing. Deuteronomy 31 itself is compositionally layered, and assignments of verses 1–13 to particular editions vary.

The corpus results cannot choose among detailed redaction models. They do support a more limited claim: Deuteronomy 31:10–13 is thoroughly integrated into the book's program of centralized covenant instruction and intergenerational Torah transmission.

The command also does not prove that a septennial national ceremony was regularly performed in the form described. Prescriptive texts may project ideals, authorize innovations by Mosaic attribution, reshape remembered practice, or formalize an institution. Joshua's narrative fulfillment is internal literary evidence, not external corroboration.

## 16.7 - Why Sukkot after the sabbatical year?<a class="anchor" id="bullet16x7"></a>
##### [Back to ToC](#TOC)

The chosen moment joins several symbolic and practical horizons:

- pilgrimage to the chosen place already gathers a wide population;
- the agricultural year has turned;
- the sabbatical cycle foregrounds land, release, dependence, and YHWH's ownership;
- public reading renews the normative order for the next seven-year cycle.

The `טף` embody the future across that cycle. Many will mature before the next assembly. Verse 13 makes precisely this temporal bridge explicit.

# 17 - Conclusions<a class="anchor" id="bullet17"></a>
##### [Back to ToC](#TOC)

1. `טף` is collective and socially functional. The 42 BHSA occurrences favor “dependents / young dependents / those unable to march” over a narrowly defined age class.
2. Deuteronomy 31:12 is a marked reuse of the term. A category usually exposed in movement, war, capture, and care is incorporated into a covenant-learning assembly.
3. The syntax gives `טף` full audience membership. Men, women, `טף`, and resident alien form the object of “assemble”; the purpose chain addresses the plural collective.
4. Verse 13 supplies the intergenerational logic. Children who “have not known” hear, learn, and acquire covenant fear for life in the land.
5. The rare `טף + גר` triad is literarily significant. It occurs only in Deuteronomy 29:10, Deuteronomy 31:12, and Joshua 8:35, joining covenant constitution, periodic Torah reading, and narrated fulfillment.
6. The passage constructs an unusually broad covenant public. Inclusion in ritual hearing does not erase social hierarchy, but it refuses to identify covenant Israel only with autonomous adult males.
7. Historical performance remains unproven. The text is strong evidence for a scribal-theological program of public Torah and collective identity; it is not by itself proof of a continuously observed early institution.

## 17.1 - Compact answer to the research question<a class="anchor" id="bullet17x1"></a>
##### [Back to ToC](#TOC)

The explicit `טף` in Deuteronomy 31:12 makes the covenant assembly coextensive with the whole lived community, including those who cannot yet function as independent, military, literate, or juridical actors. Historically and literarily, their presence turns public Torah reading into a mechanism of intergenerational identity formation. The term's wider distribution adds force to the command: the dependents whom narratives normally carry, protect, capture, feed, or endanger are here made indispensable hearers within the imagined covenant public.

## 18.1 - Limitations<a class="anchor" id="bullet18x1"></a>
